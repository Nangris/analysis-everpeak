
### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

```python

# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
# errors='coerce' convierte en NaT las fechas que no se puedan leer, en lugar de detener el código
traffic['update_time_utc'] = pd.to_datetime(traffic['update_time_utc'], errors='coerce')
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic['update_time_utc_week_ago'], errors='coerce')

# verificar el cambio
traffic.info()
```

<div class="alert alert-block alert-success">
<b>Comentario del Revisor [1° iteración]</b> <a class="tocSkip"></a><br>
<b>Éxito</b>

Correcta conversión de las columnas temporales al tipo datetime especificando el manejo de errores.
</div>

```python
# Limpia separadores y convierte columnas numéricas en eco
# Nota: ejecuta esta celda una sola vez; si la repites, quitaría el punto decimal de valores ya convertidos.
eco['city_gdp_capita'] = (eco['city_gdp_capita'].astype(str)
                          .str.replace('.', '', regex=False)    # quita separador de miles
                          .str.replace(',', '.', regex=False)   # coma decimal -> punto
                          .str.strip()
                          .astype(float))

eco['unemployment_pct'] = (eco['unemployment_pct'].astype(str)
                           .str.replace('%', '', regex=False)   # quita el símbolo %
                           .str.replace(',', '.', regex=False)
                           .str.strip()
                           .astype(float))

eco['population_m'] = (eco['population_m'].astype(str)
                       .str.replace(',', '.', regex=False)
                       .str.strip()
                       .astype(float))

# PM2.5 también viene como texto con coma decimal; se convierte para que el dataset final quede limpio
eco['pm25'] = pd.to_numeric(eco['pm25'].astype(str).str.replace(',', '.', regex=False).str.strip(),
                            errors='coerce')

# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco['population_m'] * 1000000

# verificar el cambio
eco.info()
eco.head(3)
```

## 🧩Paso 5: Unir movilidad y economía

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_hist_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Estandarizar los nombres de ciudad en ambas tablas para que coincidan:
# sin espacios extra, en minúsculas y con guiones (ej. 'buenos-aires')
traffic_2024_small['city'] = traffic_2024_small['city'].str.strip().str.lower().str.replace(' ', '-', regex=False)
eco_2024_small['city'] = eco_2024_small['city'].str.strip().str.lower().str.replace(' ', '-', regex=False)

# Unir datasets: inner join para conservar solo ciudades y años presentes en ambas tablas
merged = pd.merge(traffic_2024_small, eco_2024_small, on=['city', 'year'], how='inner')

# Mostrar las primeras 5 filas
merged.head()
```

Analizar visualmente la distribución y la relación entre indicadores de tráfico y economía en 2024, para identificar posibles patrones o tendencias generales entre ambas variables.

    - **Boxplot** → para observar la media, mediana y detectar valores atípicos.
- Visualiza la distribución de la **economía** (`city_gdp_capita`) mediante:

```python

# Crear histograma para ver la distribución de la economía (city_gdp_capita)
mean_gdp = merged['city_gdp_capita'].mean()

merged['city_gdp_capita'].hist(bins=10, figsize=(8, 4))
plt.axvline(mean_gdp, color='red', linestyle='--', label=f'Promedio: {mean_gdp:,.0f} USD')
plt.title('Distribución del PIB per cápita por ciudad (2024)')
plt.xlabel('PIB per cápita (USD)')
plt.ylabel('Número de ciudades')
plt.legend()
plt.show()
```

```python

# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
# Las dos variables tienen escalas muy distintas (minutos vs. dólares), así que el eje Y
# va en escala logarítmica para que las dos barras se vean. Las ciudades van ordenadas por PIB.
(merged.sort_values('city_gdp_capita')
       .plot(x='city', y=['jams_delay', 'city_gdp_capita'], kind='bar',
             logy=True, figsize=(14, 5)))
plt.title('JamsDelay promedio y PIB per cápita por ciudad (2024), ordenado de menor a mayor PIB')
plt.xlabel('Ciudad')
plt.ylabel('Valor (escala logarítmica)')
plt.xticks(rotation=90)
plt.show()
```

### 6.2 Comparación de congestión y PIB por ciudad

Para apoyar las conclusiones y el resumen ejecutivo, esta tabla ordena todas las ciudades unidas de mayor a menor retraso promedio. También revisa cuántas ciudades y países quedaron después de la unión, calcula la correlación entre `jams_delay` y `city_gdp_capita` y muestra las ciudades que combinan **retraso por encima de la mediana** con **PIB per cápita por debajo de la mediana**.

```python
# Tabla de apoyo: todas las ciudades unidas, de mayor a menor retraso promedio
comparacion = (merged[['city', 'country', 'jams_delay', 'traffic_index_live', 'mins_delay',
                       'city_gdp_capita', 'unemployment_pct', 'population']]
               .sort_values('jams_delay', ascending=False)
               .reset_index(drop=True))
display(comparacion)

# Cobertura de la unión
print('Ciudades en el análisis:', merged['city'].nunique())
print('Países en el análisis:', merged['country'].nunique())
print('Ciudades de eco 2024 sin datos de tráfico:', sorted(set(eco_2024_small['city']) - set(merged['city'])))

# Correlación entre retraso por congestión y PIB per cápita
print('Correlación entre jams_delay y PIB per cápita:',
      round(merged['jams_delay'].corr(merged['city_gdp_capita']), 2))

# Ciudades con retraso por encima de la mediana y PIB per cápita por debajo de la mediana
alta_congestion = merged['jams_delay'] > merged['jams_delay'].median()
bajo_pib = merged['city_gdp_capita'] < merged['city_gdp_capita'].median()
prioritarias = merged.loc[alta_congestion & bajo_pib, ['city', 'jams_delay', 'city_gdp_capita']]
display(prioritarias.sort_values('jams_delay', ascending=False))
```
